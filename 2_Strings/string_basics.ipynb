```python
#Partition

# if delimiter is found, it returns a tuple of 3 elements: (part before delimiter, delimiter, part after delimiter)

text1 = "python-spark:aws"
text1 = text1.partition("-")
print(text1)

# if delimiter is not found, it returns a tuple of 3 elements: (original string, '', '')
text2 = "python:spark:aws"
text2 = text2.partition("-")
print(text2)


# rpartition - splits the string at the last occurrence of the delimiter

text = "Python-Spark-AWS"

print(text.rpartition("-"))

```

```python
### Find duplicate words in a sentence.

text = "python spark python kafka spark python"

from collections import Counter

# Split the sentence into individual words
words = text.split()
print(words)

# Method 1: Using Counter
word_counts = Counter(words)
print(word_counts)

print(word_counts.items())

duplicates = []
for word, count in word_counts.items():
    print(f"'{word}' appears {count} times.")
    if count > 1:
        duplicates.append(word)

print(duplicates)         

    # if count > 1:
    #     print(f"'{word}' appears {count} times.")
```
