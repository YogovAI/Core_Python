```python
#Partition

# if delimiter is found, it returns a tuple of 3 elements: (part before delimiter, delimiter, part after delimiter)

text1 = "python-spark:aws"
text1 = text1.partition("-")
print(text1)

# if delimiter is not found, it returns a tuple of 3 elements: (original string, '', '')
text2 = "python:spark:aws"
text2 = text2.partition("-")
print(text2)


# rpartition - splits the string at the last occurrence of the delimiter

text = "Python-Spark-AWS"

print(text.rpartition("-"))

```

```python
# Zfill

number = "123"

print(number.zfill(8))

# Center 

text = "Python"

print(text.center(20, "-"))

#lJust - Left-aligns the string.

text = "Python"

print(text.ljust(10, "-"))

#rJust - Right-aligns the string.

text = "Python"

print(text.rjust(10, "-"))
```

```python
#translate() and maketrans() - useful for replacing multiple characters


text = "Scala is a functional language"

table = str.maketrans({"f" : "F", "l" : "L"})

print(text.translate(table))
```
