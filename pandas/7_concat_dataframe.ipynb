```python
import pandas as pd
import numpy as np
```

```python
india_weather = pd.DataFrame({
    "city" : ["mumbai", "delhi", "bengaluru"],
    "temperature" : [32, 45, 30],
    "humidity" : [80, 60, 78]
})

india_weather
```

```python
us_weather = pd.DataFrame({
    "city" : ["new york", "chicago", "orlando"],
    "temperature" : [21, 14, 35],
    "humidity" : [68, 65, 75]
})

us_weather
```

```python
new_df = pd.concat([india_weather, us_weather])
new_df
```

### Ignoring Index

```python
new_df = pd.concat([india_weather, us_weather], ignore_index=True)
new_df
```

### Concatenating to dataframe to columns level instead of row level

```python
new_df = pd.concat([india_weather, us_weather], axis=1)
new_df
```

### using index in concat dataframe

```python
new_df = pd.concat([india_weather, us_weather], keys= ["india", "us"])
new_df
```

```python
new_df.loc["us"]
```

```python
new_df = pd.concat([india_weather, us_weather], axis=1, keys=["india", "us"])
new_df
```

```python
new_df["us"]
```

### Concatenation Using Index

```python
temperature_df = pd.DataFrame({
    "city": ["delhi","mumbai","chennai"],
    "temperature": [32,45,30],
}, index=[0,1,2])
temperature_df
```

```python
windspeed_df = pd.DataFrame({
    "city": ["delhi","chennai"],
    "windspeed": [7,12],
}, index=[0,2])
windspeed_df
```

```python
new_df = pd.concat([temperature_df, windspeed_df], axis=1)
new_df
```

### Concatenate dataframe with series

```python
s = pd.Series(["Humid","Dry","Rain"], name="event")
s
```

```python
new_df = pd.concat([temperature_df, s], axis = 1)
new_df
```
