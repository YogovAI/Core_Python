```python
import pandas as pd
import numpy as np
```

## 

### Creating Dataframe using Dictionary

```python
dict1 = {"name" : ["Yogov", "Dhoni", "Tata", "Steve"],
         "Age" : [32, 40, 60, 65],
         "Profession" : ["AIEngineer", "Cricket", "Entrepeneur", "Founder"]
         }

df1 = pd.DataFrame(dict1)
df1
```

### Creating Dataframe using List and Tuples

```python
tuple1= [("Yogov", 33, "AIEngineer"), ("Dhoni", 40, "Cricket"), ("Tata", "60", "Entrepeneur"), ("Steve", 65, "Founder")]

df2 = pd.DataFrame(tuple1, columns=["Name", "Age", "Profession"])

df2
```

### Creating Dataframe using List and Dictionary

```python
list_dict1= [
        {"Name" : "Yogov", "Age" : 33, "Profession" : "AIEngineer"}, 
        {"Name" : "Dhoni", "Age" : 40, "Profession" : "Cricket"}, 
        {"Name" : "Tata",  "Age" : 60, "Profession" : "Entrepeneur"}, 
        {"Name" : "Steve", "Age" : 65, "Profession" : "Founder"}
         ]

df3 = pd.DataFrame(list_dict1)
df3
```
