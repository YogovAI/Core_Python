### Pivot Table

```python
df = pd.read_csv("weather2.csv")
df
```

```python
pivot_tbl_df = df.pivot_table(index = 'date', columns='city', values=["temperature", "humidity"])
pivot_tbl_df
```

```python
pivot_tbl_df = df.pivot_table(index = 'date', columns='city', values=["temperature", "humidity"], margins=True)
pivot_tbl_df
```

```python
pivot_tbl_df = df.pivot_table(index = 'date', columns='city', values=["temperature"], aggfunc="count")
pivot_tbl_df
```

### Grouper

```python
df = pd.read_csv("weather3.csv", parse_dates=['date'])
df
```
