```python
import pandas as pd
from pandasql import sqldf # type: ignore
```

```python
# Create a Pandas DataFrame
df = pd.DataFrame({
    'id': [1, 2, 3],
    'name': ['Alice', 'Bob', 'Charlie']
})

df
```

```python
# Define a SQL query
query = "SELECT * FROM df WHERE id > 1"
```

```python
# Execute SQL query on DataFrame
result = sqldf(query, locals())

# Print the result
print(result)
print(type(result))
result
```
